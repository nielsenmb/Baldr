# Fitting an empirical prior

This example fits a non-Gaussian sample once, checks the interpolated distribution, and reuses the fitted grid with JAX.

```python
import numpy as np

from baldr.empirical import fit_empirical
```

```python
rng = np.random.default_rng(42)
samples = np.concatenate((
    rng.normal(-1.0, 0.4, 300),
    rng.normal(1.2, 0.7, 700),
))
prior = fit_empirical(samples, grid_size=1024)
```

```python
probabilities = np.array([0.05, 0.5, 0.95])
quantiles = prior.ppf(probabilities)
np.column_stack((probabilities, quantiles, prior.cdf(quantiles)))
```

The same fitted grid can be transferred to JAX without re-running the KDE. Compile a larger prior transform in real inference code rather than only this isolated method.

```python
import jax
import jax.numpy as jnp

jax_prior = prior.grid.to_backend("jax")
compiled_ppf = jax.jit(jax_prior.ppf)
compiled_ppf(jnp.asarray(probabilities))
```
